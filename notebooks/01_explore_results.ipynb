# 01 — Explore benchmark results

Open `results/runs/results.parquet` (or the JSON sidecar) and look at the headline numbers per configuration.

```python
import json
from pathlib import Path

import pandas as pd

PARQUET = Path('results/runs/results.parquet')
JSON_SIDECAR = Path('results/runs/results.json')

if JSON_SIDECAR.is_file():
    runs = json.loads(JSON_SIDECAR.read_text(encoding='utf-8'))
    df = pd.json_normalize(runs)
else:
    df = pd.read_parquet(PARQUET)

df.head()
```

```python
# Per-config headline numbers
summary = df.groupby('config_id').agg(
    wer_mean=('quality.wer', 'mean'),
    rtfx_mean=('latency.rtfx', 'mean'),
    p95_mean=('latency.p95_latency_s', 'mean'),
    ram_peak_max=('resources.ram_peak_mb', 'max'),
    temp_peak_max=('resources.temperature_peak_c', 'max'),
    runs=('run_id', 'count'),
).round(3)
summary
```
