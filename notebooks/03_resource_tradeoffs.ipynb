# 03 — Resource trade-offs

Peak RAM, peak GPU memory, and peak temperature per configuration. The values come from tegrastats on Jetson, or from psutil in demo mode (in which case the GPU/temperature columns will be `NaN`).

```python
import json
from pathlib import Path

import pandas as pd

PARQUET = Path('results/runs/results.parquet')
JSON_SIDECAR = Path('results/runs/results.json')

if JSON_SIDECAR.is_file():
    runs = json.loads(JSON_SIDECAR.read_text(encoding='utf-8'))
    df = pd.json_normalize(runs)
else:
    df = pd.read_parquet(PARQUET)
ram = df.groupby('config_id')['resources.ram_peak_mb'].max()
gpu = df.groupby('config_id')['resources.gpu_mem_peak_mb'].max()
temp = df.groupby('config_id')['resources.temperature_peak_c'].max()
out = pd.concat([ram, gpu, temp], axis=1).round(1)
out.columns = ['peak_ram_mb', 'peak_gpu_mem_mb', 'peak_temp_c']
out
```
