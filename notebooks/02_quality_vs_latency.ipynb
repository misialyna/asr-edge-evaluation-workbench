# 02 — Quality vs real-time performance

The trade-off frontier: WER on the y-axis, RTFx on the x-axis. Configurations to the right are faster; lower is better on the y-axis.

```python
import json
from pathlib import Path

import pandas as pd

PARQUET = Path('results/runs/results.parquet')
JSON_SIDECAR = Path('results/runs/results.json')

if JSON_SIDECAR.is_file():
    runs = json.loads(JSON_SIDECAR.read_text(encoding='utf-8'))
    df = pd.json_normalize(runs)
else:
    df = pd.read_parquet(PARQUET)
df_ok = df[df['quality.wer'].notna()]
df_ok[['config_id', 'quality.wer', 'latency.rtfx', 'audio.file_id']].head()
```

```python
import plotly.express as px

fig = px.scatter(
    df_ok,
    x='latency.rtfx',
    y='quality.wer',
    color='config_id',
    hover_data=['audio.file_id'],
    title='Quality vs Real-Time Performance',
)
fig.add_vline(x=1.0, line_dash='dash', line_color='grey', annotation_text='RTFx=1')
fig.show()
```
